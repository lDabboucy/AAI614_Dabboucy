```python
# Read the data from Github

import pandas as pd
url =  "https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%203/niaaa-report2009.csv"

df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
for column in df.select_dtypes(include="number"):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    
    print(f"{column}: Q1 = {Q1:.3f}, Q3 = {Q3:.3f}, IQR = {IQR:.3f}, Lower = {lower_limit:.4f}, Upper = {upper_limit:.4f}")
    outliers = df [
        (df[column] < lower_limit) | (df[column] > upper_limit)
        ]
    if outliers.empty:
        print("No Outliers")
    else:
        display(outliers[["State", column]])
    print()
```

```python
import matplotlib.pyplot as plt

df.select_dtypes(include="number").boxplot()
plt.title("Alcohol Consumption by Type")
plt.ylabel("Per capita consumption")
plt.show()
```

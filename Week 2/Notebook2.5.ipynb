```python
import pandas as pd
import time
import ssl


ssl._create_default_https_context = ssl._create_unverified_context
```

```python
class Timer:
    def __enter__(self):
        self.start = time.perf_counter()
        return self

    def __exit__(self, *args):
        self.end = time.perf_counter()
        self.interval = self.end - self.start
```

```python
small_file_path_CSV = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Small.csv'
medium_file_path_CSV = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Medium.csv'
large_file_path_CSV = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Large.csv'
```

```python
# Read CSV files and time them
results = []

with Timer() as t_pd:
    df = pd.read_csv(small_file_path_CSV)

small_csv_time = t_pd.interval
results.append(["Small", "CSV", small_csv_time])
print("Small CSV File: ", small_csv_time)



with Timer() as t_pd:
    df = pd.read_csv(medium_file_path_CSV)

medium_csv_time = t_pd.interval
results.append(["Small", "CSV", medium_csv_time])
print("Medium CSV File: ", medium_csv_time)

with Timer() as t_pd:
    df = pd.read_csv(large_file_path_CSV)

large_csv_time = t_pd.interval
results.append(["Small", "CSV", large_csv_time])
print("Large CSV File: ", large_csv_time)
```

##### Read the Parquet files and time them




```python
small_file_path_Parquet = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Small.parquet'
medium_file_path_Parquet = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Medium.parquet'
large_file_path_Parquet  = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Large.parquet'
```

```python
with Timer() as t_pd:
    df = pd.read_parquet(small_file_path_Parquet )

small_parquet_time = t_pd.interval
results.append(["Small", "Parquet", small_parquet_time])
print("Small Parquet File: ",small_parquet_time)

with Timer() as t_pd:
    df = pd.read_parquet(medium_file_path_Parquet )

medium_parquet_time = t_pd.interval
results.append(["Small", "Parquet", medium_parquet_time])
print("Medium Parquet File: ",medium_parquet_time)

with Timer() as t_pd:
    df = pd.read_parquet(large_file_path_Parquet )
print("Large Parquet File: ",t_pd.interval)
```

#### Read the zip files and time them

```python
small_file_path_zip = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Small.zip'
medium_file_path_zip = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Medium.zip'
large_file_path_zip  = 'https://raw.githubusercontent.com/lDabboucy/AAI614_Dabboucy/main/Week%202/data/niaaa-report2009_Large.zip'
```

```python

with Timer() as t_pd:
    df = pd.read_csv(small_file_path_zip,compression="zip" )
print("Small zip File: ",t_pd.interval)

with Timer() as t_pd:
    df = pd.read_csv(medium_file_path_zip,compression="zip" )
print("Medium zip File: ",t_pd.interval)

with Timer() as t_pd:
    df = pd.read_csv(large_file_path_zip,compression="zip" )
print("Large zip File: ",t_pd.interval)
```
